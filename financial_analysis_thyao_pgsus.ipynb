```python
import pandas as pd 
import matplotlib.pyplot as plt

df = pd.read_csv("financial_ratios.csv")

thy = df[df["COMPANY"]=="THYAO"].sort_values("YEAR")
pgs = df[df["COMPANY"]=="PGSUS"].sort_values("YEAR")

fig, ax = plt.subplots(1,2, figsize=(12,4))

ax[0].plot(thy["YEAR"],thy["NET MARGİN"],marker="o",label="THYAO")
ax[0].plot(pgs["YEAR"],pgs["NET MARGİN"],marker="o",label="PGSUS")
ax[0].set_title("Net Margin")
ax[0].grid(alpha=0.3)
ax[0].legend()

ax[1].plot(thy["YEAR"],thy["ROE"],marker="o",label="THYAO")
ax[1].plot(pgs["YEAR"],pgs["ROE"],marker="o",label="THYAO")
ax[1].set_title("ROE COMPARİSON")
ax[1].grid(alpha=0.3)
ax[1].legend()

plt.tight_layout()
plt.show()
```
